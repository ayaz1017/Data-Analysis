```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df=pd.read_csv('Sales.csv',encoding='unicode_escape')
```

```python
df
```

```python
print("Shape of the data:",df.shape)
```

```python
df.head(3)
```

```python
df.info()
```

```python
df.drop(['Status','unnamed1'],axis=1,inplace=True)
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
#chnage the data type of amount
df['Amount']=df['Amount'].astype('int')
```

```python
df.describe()
```

```python
df['Gender'].value_counts()
```

```python
plt.figure(figsize=(7,4))
sns.countplot(x=df['Gender'],color='black')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.style.use('seaborn-v0_8-whitegrid')
ax = sns.countplot(
    data=df,
    x='Age Group',
    hue='Gender',
    palette=['#FF6B6B', '#4ECDC4'],
    alpha=0.9
)
plt.title('Distribution of Age Groups by Gender', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Age Group', fontsize=12, labelpad=10)
plt.ylabel('Count', fontsize=12, labelpad=10)
for container in ax.containers:
    ax.bar_label(container, fmt='%d', fontsize=10, fontweight='bold')
plt.legend(title='Gender', title_fontsize=12, fontsize=10, loc='upper right')
sns.despine()
ax.set_facecolor('#f8f9fa')
plt.gcf().set_facecolor('#f8f9fa')
plt.tight_layout(pad=3)
plt.show()
```

```python
sales_age = df.groupby(['Age Group'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

sns.barplot(x = 'Age Group',y= 'Amount' ,data = sales_age)
```

```python
# total number of orders from top 10 states
sales_state = df.groupby(['State'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)
plt.figure(figsize=(15, 6))
plt.style.use('seaborn-v0_8')
ax = sns.barplot(
    data=sales_state,
    x='State',
    y='Orders',
    palette='Blues_d'
)
for i, bar in enumerate(ax.patches):
    ax.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height() + 0.3,
        f'{int(bar.get_height())}',
        ha='center', fontsize=10
    )
plt.title('Top 10 States by Order Volume', fontsize=14, pad=10)
plt.xlabel('State', fontsize=12)
plt.ylabel('Total Orders', fontsize=12)
plt.xticks(rotation=15)
sns.despine()
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# total amount/sales from top 10 states
sales_state = df.groupby(['State'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)

sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(data = sales_state, x = 'State',y= 'Amount')
```
